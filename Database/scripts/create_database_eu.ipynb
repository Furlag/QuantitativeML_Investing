## Collect data from yahoo finance to the data dock

```python
from data_collector.yfinance.collector import download_data


tickers = (
    ['EXSA.DE']  # ETF: iShares STOXX Europe 600 UCITS ETF (Tracks the STOXX600)
    + tickers['ticker'].tolist()
)


download_data(
    tickers,
    start="2000-01-01", end="2099-12-31",

    append_index_data=True,
    append_index_data__index_name="EXSA.DE",
    append_index_data__features=['close'],

    target_dir=qlib_dir / "_raw_data_from_yfinance"
)
```

## Dump data from the data dock to qlib database format

```python
from dump_bin import DumpDataAll, DumpDataUpdate, DumpDataFix


_dump = DumpDataAll(
    data_path=qlib_dir / "_raw_data_from_yfinance",
    qlib_dir=qlib_dir,
    freq="day",
    max_workers=4,
)
_dump.dump()
```
