```python
from scripts.download import yahoo_downloader

import pathlib
import os
import pandas as pd


# Read tickers
T = pd.read_csv('C:/Projects/Investing/Quantitative/Database/Ticker_Lookup.csv', skiprows=1)
T.head()
```

```python
###################
# Spain
###################
tickers = (
    ['^IBEX', 'INDC.MC']
    + T['^IBEX'].dropna().tolist() + T['INDC.MC'].dropna().tolist()
)

qlib_dir = os.path.join(os.path.dirname(os.path.abspath('__file__')), "Qdata", "SPAIN")
yahoo_downloader(
    tickers,
    qlib_dir=qlib_dir, dump_mode="dump_all",
    start="2000-01-01", end="2023-01-01",
    append_index_data='^IBEX',
    python_path="C:/Projects/Investing/Quantitative/.invest_venv/Scripts/python"
)  # dump_all / dump_fix
```
